```python
# fast_dga_analysis_fixed.py

import pandas as pd
import matplotlib.pyplot as plt

# Load dataset (3 columns: label,family,domain)
df = pd.read_csv("dga-dataset.txt", names=["label", "family", "domain"])

# Count benign vs malicious
counts = df["label"].value_counts()

# Print results
total = counts.sum()
print("Domain counts by class:")
for label, count in counts.items():
    percentage = (count / total) * 100
    print(f"{label.capitalize()}: {count} ({percentage:.2f}%)")

# Plot charts
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

# Bar chart
counts.plot(kind="bar", color=["green", "red"], ax=axes[0])
axes[0].set_title("Benign vs Malicious Domain Counts")
axes[0].set_xlabel("Class")
axes[0].set_ylabel("Number of Domains")

# Add percentage labels
for i, (label, count) in enumerate(counts.items()):
    percentage = (count / total) * 100
    axes[0].text(i, count, f"{percentage:.2f}%", ha="center", va="bottom", fontsize=9)

# Pie chart
counts.plot(kind="pie", autopct="%1.2f%%", colors=["green", "red"], ax=axes[1])
axes[1].set_ylabel("")
axes[1].set_title("Benign vs Malicious Ratio")

plt.tight_layout()
plt.show()

```

```python
# malware_family_analysis.py

import pandas as pd
import matplotlib.pyplot as plt

# Load dataset (3 columns: label,family,domain)
df = pd.read_csv("dga-dataset.txt", names=["label", "family", "domain"])

# Filter only malicious domains (label = dga)
malicious_df = df[df["label"] == "dga"]

# Count malware families
family_counts = malicious_df["family"].value_counts()

# Print results
total = family_counts.sum()
print("Malware family distribution:")
for family, count in family_counts.items():
    percentage = (count / total) * 100
    print(f"{family}: {count} ({percentage:.2f}%)")

# Plot charts
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Bar chart
family_counts.plot(kind="bar", ax=axes[0], color="red")
axes[0].set_title("Malware Family Counts (DGA domains)")
axes[0].set_xlabel("Family")
axes[0].set_ylabel("Number of Domains")

# Add percentage labels on bars
for i, (family, count) in enumerate(family_counts.items()):
    percentage = (count / total) * 100
    axes[0].text(i, count, f"{percentage:.2f}%", ha="center", va="bottom", fontsize=8, rotation=90)



# Pie chart with better label placement
family_counts.plot(
    kind="pie",
    autopct="%1.1f%%",        # show 1 decimal (e.g., 4.0%)
    startangle=90,
    colormap="tab20",
    ax=axes[1],
    textprops={'fontsize': 8},  # smaller font for better readability
    pctdistance=0.75,           # move percentage labels inside
    labeldistance=1.1           # push family names slightly outward
)

axes[1].set_ylabel("")
axes[1].set_title("Malware Family Distribution (%)")


plt.tight_layout()
plt.show()
```

```python


import pandas as pd
import matplotlib.pyplot as plt

# Load dataset (3 columns: label, family, domain)
df = pd.read_csv(
    "dga-dataset.txt",
    names=["label", "family", "domain"]
)

# Count benign vs malicious
counts = df["label"].value_counts()

# Print results
total = counts.sum()
print("Domain counts by class:")

for label, count in counts.items():
    percentage = (count / total) * 100
    print(f"{label.capitalize()}: {count} ({percentage:.2f}%)")

# Assign colors based on class labels
colors = [
    "red" if label.strip().lower() in ["dga", "malicious"]
    else "green"
    for label in counts.index
]

# Plot charts
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

# Bar chart
counts.plot(kind="bar", color=colors, ax=axes[0])
axes[0].set_title("Benign vs Malicious Domain Counts")
axes[0].set_xlabel("Class")
axes[0].set_ylabel("Number of Domains")

# Add percentage labels
for i, (label, count) in enumerate(counts.items()):
    percentage = (count / total) * 100
    axes[0].text(
        i, count, f"{percentage:.2f}%",
        ha="center", va="bottom", fontsize=9
    )

# Pie chart
counts.plot(
    kind="pie",
    autopct="%1.2f%%",
    colors=colors,
    ax=axes[1]
)
axes[1].set_ylabel("")
axes[1].set_title("Benign vs Malicious Ratio")

plt.tight_layout()
plt.show()

```

```python

import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv(
    "dga-dataset.txt",
    names=["label", "family", "domain"]
)

# Clean labels and count classes
df["label"] = df["label"].astype(str).str.strip()
counts = df["label"].value_counts()
total = counts.sum()

# Print results
print("Domain counts by class:")
for label, count in counts.items():
    percentage = count / total * 100
    print(f"{label.capitalize()}: {count:,} ({percentage:.2f}%)")

# Assign colors based on labels
def get_color(label):
    label = label.strip().lower()
    if "dga" in label or "malicious" in label:
        return "#D62728"  # Red
    return "#2E8B57"      # Green

colors = [get_color(label) for label in counts.index]

# Publication-style appearance
plt.rcParams.update({
    "font.family": "DejaVu Sans",
    "font.size": 11,
    "axes.titlesize": 13,
    "axes.titleweight": "bold",
    "figure.facecolor": "white",
    "axes.facecolor": "white"
})

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# --------------------------------
# 1. Horizontal bar chart
# --------------------------------
ax = axes[0]

bars = ax.barh(
    counts.index.str.capitalize(),
    counts.values,
    color=colors,
    height=0.55
)

ax.set_title("Domain Counts by Class", pad=14)
ax.set_xlabel("Number of Domains")
ax.set_ylabel("")

# Subtle vertical gridlines
ax.set_axisbelow(True)
ax.xaxis.grid(True, linestyle="--", alpha=0.25)
ax.yaxis.grid(False)

# Remove unnecessary borders
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.spines["left"].set_visible(False)

# Add count and percentage labels
for bar, count in zip(bars, counts.values):
    percentage = count / total * 100
    ax.annotate(
        f"{count:,} ({percentage:.2f}%)",
        xy=(bar.get_width(), bar.get_y() + bar.get_height() / 2),
        xytext=(6, 0),
        textcoords="offset points",
        va="center",
        fontsize=9
    )

ax.set_xlim(0, counts.max() * 1.38)

# --------------------------------
# 2. Donut chart
# --------------------------------
ax = axes[1]

wedges, texts, autotexts = ax.pie(
    counts.values,
    labels=[label.capitalize() for label in counts.index],
    colors=colors,
    startangle=90,
    counterclock=False,
    autopct="%1.2f%%",
    pctdistance=0.72,
    wedgeprops={"width": 0.38, "edgecolor": "white",
                "linewidth": 2},
    textprops={"fontsize": 10}
)

for text in autotexts:
    text.set_color("white")
    text.set_fontweight("bold")
    text.set_fontsize(9)

# Total in the center
ax.text(
    0, 0.08, "TOTAL",
    ha="center", va="center",
    fontsize=10, color="gray"
)
ax.text(
    0, -0.12, f"{total:,}",
    ha="center", va="center",
    fontsize=14, fontweight="bold"
)

ax.set_title("Class Distribution", pad=14)

# Final layout
fig.suptitle(
    "DGA Dataset Overview",
    fontsize=15,
    fontweight="bold",
    y=1.02
)

plt.tight_layout()

# Save a high-resolution version for your article
plt.savefig(
    "dga_dataset_overview.png",
    dpi=300,
    bbox_inches="tight",
    facecolor="white"
)

plt.show()
```

```python

# malware_family_analysis.py

import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv(
    "dga-dataset.txt",
    names=["label", "family", "domain"]
)

# Filter only malicious domains
malicious_df = df[df["label"].astype(str).str.strip().str.lower() == "dga"]

# Count malware families, sorted from largest to smallest
family_counts = malicious_df["family"].value_counts()

# Print results
total = family_counts.sum()
print("Malware family distribution:")

for family, count in family_counts.items():
    percentage = (count / total) * 100
    print(f"{family}: {count:,} ({percentage:.2f}%)")

# Professional chart styling
plt.rcParams.update({
    "font.family": "DejaVu Sans",
    "font.size": 10,
    "axes.titlesize": 13,
    "axes.titleweight": "bold",
    "figure.facecolor": "white",
    "axes.facecolor": "white"
})

# Create charts
fig, axes = plt.subplots(1, 2, figsize=(15, 7))

# Consistent colors for each family
cmap = plt.get_cmap("tab20")
colors = [cmap(i % 20) for i in range(len(family_counts))]

# ------------------------------------
# 1. Horizontal bar chart
# ------------------------------------
ax = axes[0]

# Reverse order so the largest family appears at the top
plot_counts = family_counts.sort_values(ascending=True)
plot_colors = [
    cmap(i % 20)
    for i in range(len(family_counts) - 1, -1, -1)
]

bars = ax.barh(
    plot_counts.index.astype(str),
    plot_counts.values,
    color=plot_colors,
    height=0.72
)

ax.set_title("Malware Family Counts", pad=12)
ax.set_xlabel("Number of DGA Domains")
ax.set_ylabel("")

# Light gridlines and clean borders
ax.set_axisbelow(True)
ax.xaxis.grid(True, linestyle="--", alpha=0.25)
ax.yaxis.grid(False)
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.spines["left"].set_visible(False)

# Show counts and percentages at the end of each bar
for bar, count in zip(bars, plot_counts.values):
    percentage = count / total * 100

    ax.annotate(
        f"{count:,} ({percentage:.1f}%)",
        xy=(bar.get_width(), bar.get_y() + bar.get_height() / 2),
        xytext=(5, 0),
        textcoords="offset points",
        ha="left",
        va="center",
        fontsize=8
    )

ax.set_xlim(0, family_counts.max() * 1.38)

# ------------------------------------
# 2. Donut chart
# ------------------------------------
ax = axes[1]

wedges, _ = ax.pie(
    family_counts.values,
    startangle=90,
    counterclock=False,
    colors=colors,
    wedgeprops={
        "width": 0.38,
        "edgecolor": "white",
        "linewidth": 1.5
    }
)

# Display total number in the center
ax.text(
    0, 0.08, "TOTAL DGA",
    ha="center", va="center",
    fontsize=10, color="gray"
)
ax.text(
    0, -0.12, f"{total:,}",
    ha="center", va="center",
    fontsize=15, fontweight="bold"
)

ax.set_title("Malware Family Distribution", pad=12)

# Legend includes each family's count and percentage
legend_labels = [
    f"{family}: {count:,} ({count / total * 100:.1f}%)"
    for family, count in family_counts.items()
]

ax.legend(
    wedges,
    legend_labels,
    title="Malware Families",
    loc="center left",
    bbox_to_anchor=(0.95, 0.5),
    fontsize=8,
    title_fontsize=9,
    frameon=False
)

# Final layout
fig.suptitle(
    "DGA Malware Family Analysis",
    fontsize=16,
    fontweight="bold",
    y=0.99
)

plt.tight_layout(rect=[0, 0, 0.88, 0.95])

# Save high-resolution figure for academic use
plt.savefig(
    "malware_family_analysis.png",
    dpi=300,
    bbox_inches="tight",
    facecolor="white"
)

plt.show()
```
